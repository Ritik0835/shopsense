```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../data/shopsense.db")

def q(sql):
    return pd.read_sql(sql, conn)

q("SELECT COUNT(*) AS n FROM products")
```

 Rows per split

```python
q("""
SELECT split, COUNT(*) AS n
FROM products
GROUP BY split
""")
```

Price stats per main category (train only)

```python
q("""
SELECT main_category,
       COUNT(*)            AS n,
       ROUND(AVG(price),2) AS avg_price,
       MIN(price)          AS min_price,
       MAX(price)          AS max_price
FROM products
WHERE split = 'train'
GROUP BY main_category
ORDER BY avg_price DESC
""")
```

Top 10 priciest sub-categories (at least 50 products)

```python
q("""
SELECT sub_category,
       COUNT(*)            AS n,
       ROUND(AVG(price),2) AS avg_price
FROM products
WHERE split = 'train'
GROUP BY sub_category
HAVING COUNT(*) >= 50
ORDER BY avg_price DESC
LIMIT 10
""")
```

The 10 most expensive products

```python
q("""
SELECT "Product Name", main_category, price
FROM products
ORDER BY price DESC
LIMIT 10
""")
```

Very short descriptions

```python
q("""
SELECT COUNT(*) AS short_descriptions
FROM products
WHERE desc_length < 20
""")
```
