```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("../data/processed/products_train_clean.csv")
df["log_price"] = np.log1p(df["price"])

print("skew of price:    ", round(df["price"].skew(), 2))
print("skew of log price:", round(df["log_price"].skew(), 2))
```

```python
a = df.loc[df["main_category"] == "Toys & Games", "log_price"]
b = df.loc[df["main_category"] == "Home & Kitchen", "log_price"]

t, p_t = stats.ttest_ind(a, b, equal_var=False)
u, p_u = stats.mannwhitneyu(a, b)

print("t-test p-value:      ", p_t)
print("Mann-Whitney p-value:", p_u)
print("Typical price ratio: ", round(np.exp(a.mean() - b.mean()), 2))
```

```python
toys = df[df["main_category"] == "Toys & Games"]
counts = toys["sub_category"].value_counts()
big = counts[counts >= 100].index

groups = [toys.loc[toys["sub_category"] == s, "log_price"] for s in big]
h, p_k = stats.kruskal(*groups)
print("Groups tested:", len(groups))
print("Kruskal-Wallis p-value:", p_k)
```

```python
rho, p_s = stats.spearmanr(df["desc_length"], df["price"])
print("Spearman correlation:", round(rho, 3), " p-value:", p_s)
```
